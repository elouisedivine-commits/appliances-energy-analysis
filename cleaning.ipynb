##Cleaning Log
- Dataset has 19,735 rows and 29 columns. This matches the dataset page.
- the 'date' is being stores as text or str, this needs converting to a datetime.
- All the other columns are numeric (int and float).
- Missing Values: We have checked all 29 columns for missing values and found none, so no filling or dropping was needed.
- Duplicate Rows: There are no repeated rows.

##Part 2:Fixes
- Converted 'date' from text to datetime. The data runs from 2016-01-11 17:00 to 2016-05-27 18:00 (137 days)
- Created `hour` (0 to 23) from the timestamp, and `day_night`: Day = 6:00 am to 9:59 pm, Night = 10:00 pm to 5:59 am. This gives 13,159 Day readings and 6,576 Night readings (19,735 total). The groups are unequal in size because Day covers 16 hours and Night covers 8, which we'll mention when checking the t-test's assumptions.
- Dropped `rv1` and `rv2` because they are random-number columns with no real meaning, so they would only add noise. 
- The dataset now has 19,735 rows and 29 columns (27 original columns plus `hour` and `day_night`).

```python
df["date"] = pd.to_datetime(df["date"])
df["date"].dtype
```

```python
df["date"].min(), df["date"].max()
```

```python
df["hour"] = df["date"].dt.hour
df["day_night"] = df["hour"].apply(lambda h: "Day" if 6 <= h < 22 else "Night")

df[["date", "hour", "day_night"]].head()
```

```python
df["day_night"].value_counts()
```

```python
df = df.drop(columns=["rv1", "rv2"])
df.shape
```

```python
df.to_csv("energydata_clean.csv", index=False)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 4))
plt.hist(df["Appliances"], bins=50)
plt.title("Distribution of Appliance Energy Use")
plt.xlabel("Appliance energy use (Wh per 10-minute reading)")
plt.ylabel("Number of readings")
plt.show()
```

### Part 3: Exploration
- Appliances: mean 97.7 Wh, median 60 Wh, std 102.5, min 10, max 1,080. The mean is well above the median, and the histogram is strongly right-skewed: most readings are low, with a long tail of high-use readings.

```python
q1 = df["Appliances"].quantile(0.25)
q3 = df["Appliances"].quantile(0.75)
iqr = q3 - q1
lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

outliers = df[(df["Appliances"] < lower) | (df["Appliances"] > upper)]
print("Lower fence:", lower)
print("Upper fence:", upper)
print("Outlier rows:", len(outliers))
print("Share of data (%):", round(100 * len(outliers) / len(df), 1))
```

```python
import seaborn as sns

plt.figure(figsize=(6, 4))
sns.boxplot(data=df, x="day_night", y="Appliances")
plt.title("Appliance Energy Use: Day vs. Night")
plt.xlabel("Period (Day = 6 am to 10 pm)")
plt.ylabel("Appliance energy use (Wh per 10-minute reading)")
plt.show()
```

### 4 Outliers
- Method: It uses the 1.5 × IQR rule (Q1 = 50, Q3 = 100, IQR = 50), where IQR is the distance between the 25% and 75% values, giving fences of -25 and 175Wh. No readings fall below the lower fence. 2,138 readings(10.8%) are above 175Wh. The boxplot shows most of them occur during the day.
- Decision: KEPT. They appear to be real-high use events and not actual errors, they also make up a large share to discard, and that removing them would hide the behavior we want to study. 

```python
hourly = df.groupby("hour")["Appliances"].mean()

plt.figure(figsize=(8, 4))
plt.plot(hourly.index, hourly.values, marker="o")
plt.title("Average Appliance Energy Use by Hour of Day")
plt.xlabel("Hour of day (0 = midnight)")
plt.ylabel("Average appliance energy use (Wh per 10-minute reading)")
plt.xticks(range(0, 24, 2))
plt.show()
```

##Update: 
- Average use by hour: the lowest is overnight (about 50Wh, a first peak near 11 am (about 133Wh), a small dip mid-afternoon, and the highest peak at 6 pm (about 190Wh)).
- Correlation heatmap: all correlations with Appliances are weak. Temperature columns are strongly correlated with each other (like T6 and T_out), as are humidity columns, so predictors overlap. The daily pattern is non-linear, so a single correlation understates the effect of time of day.

- Correlations with Appliances (Pearson): hour +0.22, lights +0.20, T6 +0.12, T2 +0.12, T_out +0.10, Windspeed +0.09, RH_out -0.15, RH_8 -0.09. All are weak (the strongest explains only about 5% of the variation). Temperature and humidity in individual rooms have almost no linear relationship with appliance use. Correlation does not show causation: for example, lights and appliances likely both rise when people are home. Outdoor variables may partly reflect time of day.



## Summary of cleaning decisions
- Converted the 'date' from text to datetime.
- Created 'hour' and 'day_night'. 
- Dropped 'rv1' and 'rv2' (those random columns with no real meaning)
- No missing Values or duplicate rows were found, so nothing was removed.
- Kept all the outliers (2,138 readings above 175Wh, 10.8%) because they look like real high-use events.
